```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:\\Users\\aktiw\\Downloads\\daily_sales_365.csv")
df.head()
```

```python
print("Shape:", df.shape)
print("\nSummary:")
display(df.describe().round(2))
print("\nTotal Sales:", round(df["Sales"].sum(),2))
print("Total Profit:", round(df["Profit"].sum(),2))
print("Total Orders:", int(df["Orders"].sum()))
```

```python
monthly=df.groupby("Month", sort=False).agg(
    Sales=("Sales","sum"),
    Orders=("Orders","sum"),
    Profit=("Profit","sum")
)
display(monthly.round(2))
```

```python
plt.figure(figsize=(12,5))
plt.plot(df["Date"], df["Sales"])
plt.title("Daily Sales Trend — 2025")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))
monthly["Sales"].plot(kind="bar")
plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))
monthly["Orders"].plot(kind="bar")
plt.title("Monthly Orders Comparison")
plt.xlabel("Month")
plt.ylabel("Total Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df["Date"], df["Profit"])
plt.title("Daily Profit Trend")
plt.xlabel("Date")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Orders"], df["Sales"], alpha=0.5)
plt.title("Orders vs Sales")
plt.xlabel("Orders")
plt.ylabel("Sales")
plt.show()
```
